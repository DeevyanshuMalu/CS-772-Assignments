```python
from nltk.corpus import brown
from nltk import download
import random
import pickle
import os

random.seed(42)

download("brown")
download("universal_tagset")

sentences = brown.tagged_sents(tagset="universal")
sentences = [[(w.lower(), t) for w, t in sent] for sent in sentences]
random.shuffle(sentences)
total_sentences = len(sentences)
```

```python
train_frac = 0.8
val_frac = 0.1
```

```python
with open("train_data.pkl", "wb") as f:
	pickle.dump(sentences[:int(total_sentences * train_frac)], f)

with open("val_data.pkl", "wb") as f:
	pickle.dump(sentences[int(total_sentences * train_frac):int(total_sentences * (train_frac + val_frac))], f)

with open("test_data.pkl", "wb") as f:
	pickle.dump(sentences[int(total_sentences * (train_frac + val_frac)):], f)
```
